```python
api_url = "https://id.wikipedia.org/w/api.php"

params = {
    "action": "query",
    "list": "search",
    "srsearch": "teknologi",
    "srnamespace": 0,
    "srlimit": 200,
    "format": "json"
}

headers = {
    "User-Agent": "WebMiningStudent/1.0 (educational project)"
}

response = requests.get(
    api_url,
    params=params,
    headers=headers,
    timeout=30
)

print("Status Code:", response.status_code)

data = response.json()
hasil_pencarian = data["query"]["search"]

print("Jumlah artikel ditemukan:", len(hasil_pencarian))
```

```python
daftar_url = []

for item in hasil_pencarian:
    judul = item["title"]
    url = "https://id.wikipedia.org/wiki/" + judul.replace(" ", "_")
    
    daftar_url.append({
        "judul": judul,
        "url": url
    })

print("Jumlah URL:", len(daftar_url))
```

```python
hasil_data = []

print("Mulai proses ekstraksi...")
```

```python
for i, item in enumerate(daftar_url, start=1):
    judul = item["judul"]
    url = item["url"]

    try:
        response = requests.get(
            url,
            headers=headers,
            timeout=30
        )

        if response.status_code == 200:
            html = response.text

            teks = trafilatura.extract(
                html,
                include_comments=False,
                include_tables=False
            )

            if teks:
                jumlah_kata = len(teks.split())

                hasil_data.append({
                    "id": i,
                    "judul": judul,
                    "url": url,
                    "teks": teks,
                    "jumlah_kata": jumlah_kata
                })

                print(f"{i}/200 berhasil: {judul}")
            else:
                print(f"{i}/200 teks tidak ditemukan: {judul}")

        else:
            print(f"{i}/200 gagal ({response.status_code}): {judul}")

        time.sleep(0.5)

    except Exception as e:
        print(f"{i}/200 error: {e}")
```

```python
df = pd.DataFrame(hasil_data)

print("Jumlah data berhasil:", len(df))
```

```python
def bersihkan_teks(teks):
    teks = teks.lower()
    teks = re.sub(r'[^a-zA-ZÀ-ÿ\s]', ' ', teks)
    teks = re.sub(r'\s+', ' ', teks)
    return teks.strip()

df["teks_bersih"] = df["teks"].apply(bersihkan_teks)

print("Pembersihan teks selesai")
```

```python
semua_kata = []

for teks in df["teks_bersih"]:
    semua_kata.extend(teks.split())

kata_unik = set(semua_kata)

print("Jumlah seluruh kata:", len(semua_kata))
print("Jumlah kata unik:", len(kata_unik))
```
